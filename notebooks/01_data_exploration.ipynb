# 01 - Exploratory Data Analysis (EDA)
Comprehensive data profiling, statistical inspection, class balance evaluation, and feature correlation analysis on `data/raw/financial_fraud_detection_dataset.csv`.


```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root to sys.path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.data.ingestion import load_raw_dataset
from src.data.validation import validate_dataset_schema

sns.set_theme(style='whitegrid')
```

## 1. Dataset Dimensions & Inspection
Load raw dataset and display dimensions, column data types, and head rows.


```python
df_raw = load_raw_dataset()
print(f'Dataset Shape: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns')
df_raw.info()
```

## 2. Data Quality & Missing Value Audit
Validate schema integrity, missing value counts, and duplicate record check.


```python
validation_summary = validate_dataset_schema(df_raw)
print('Validation Result:', validation_summary['is_valid'])
print('Duplicate Records:', validation_summary['duplicate_count'])
print('Total Missing Values:', validation_summary['total_null_count'])
```

## 3. Target Class Imbalance Analysis
Analyze ratio between Genuine (0) and Fraudulent (1) transaction classes.


```python
target_counts = df_raw['Fraudulent'].value_counts()
target_pcts = df_raw['Fraudulent'].value_counts(normalize=True) * 100
print('Target Counts:\n', target_counts)
print('\nTarget Percentages (%):\n', target_pcts.round(2))

plt.figure(figsize=(6, 4))
sns.countplot(data=df_raw, x='Fraudulent', palette=['#38bdf8', '#ef4444'])
plt.title('Fraudulent vs Genuine Class Distribution')
plt.show()
```

## 4. Numerical Feature Distributions & Fraud Comparisons
Compare distributions of transaction amount, average spend, and account age across target classes.


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.boxplot(data=df_raw, x='Fraudulent', y='Transaction_Amount', ax=axes[0], palette=['#38bdf8', '#ef4444'])
axes[0].set_title('Transaction Amount by Class')

sns.boxplot(data=df_raw, x='Fraudulent', y='Average_Spend', ax=axes[1], palette=['#38bdf8', '#ef4444'])
axes[1].set_title('Average Spend by Class')

sns.boxplot(data=df_raw, x='Fraudulent', y='Account_Age_Days', ax=axes[2], palette=['#38bdf8', '#ef4444'])
axes[2].set_title('Account Age (Days) by Class')
plt.tight_layout()
plt.show()
```

## 5. Categorical Risk Breakdown
Analyze fraud rate across Merchant Category, Payment Method, and Device Type.


```python
plt.figure(figsize=(10, 4))
cat_fraud = df_raw.groupby('Merchant_Category')['Fraudulent'].mean().reset_index()
sns.barplot(data=cat_fraud, x='Merchant_Category', y='Fraudulent', palette='magma')
plt.title('Fraud Rate by Merchant Category')
plt.ylabel('Fraud Rate')
plt.xticks(rotation=30)
plt.show()
```
